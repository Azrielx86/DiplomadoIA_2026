```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Data/HollywoodMovies.csv")
df.head()
```

## Agrupación por género

```python
genres = df.groupby("Genre")["Movie"].count()

x = genres.index
y = genres.values
plt.xticks(rotation=45, ha="right")

plt.xlabel("Genre")
plt.ylabel("Movies count")

plt.bar(x, y)
plt.show()
```
