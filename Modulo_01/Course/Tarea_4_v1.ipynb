### 2007-2013 Número de Películas por Género

Hay más comedia

```python
genres = df.groupby("Genre")["Movie"].count()

x = genres.index
y = genres.values
plt.xticks(rotation=45, ha="right")

plt.xlabel("Genre")
plt.ylabel("Movies count")

plt.bar(x, y)
plt.show()
```

### Mayor ganancia por género

Las películas de acción venden más

```python
genres = df.groupby("Genre")["WorldGross"].sum()

x = genres.index
y = genres.values
plt.xticks(rotation=45, ha="right")

plt.xlabel("Genre")
plt.ylabel("Recaudación")

plt.bar(x, y)
plt.show()
```
