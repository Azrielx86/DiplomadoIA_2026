```python
import matplotlib.pyplot as plt
import numpy as np
```

# Gráfica lineal

```python
x = np.arange(0, 10)
y = x ** 2

plt.plot(x, y)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Simple plot')
plt.show()
```

# Gráfca de barras

```python
categorias = ["A", "B", "C"]
valores = [10, 15, 7]
plt.bar(categorias, valores)
plt.title(" Comparacion por categoria ")
plt.show()
```

# Gráfcas de dispersión

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.xlabel(" Variable X")
plt.ylabel(" Variable Y")

plt.show()
```

```python
import seaborn as sns

data = np.random.normal(0, 1, 1000)
sns.histplot(data, bins=30)
plt.show()
```

```python
df = sns.load_dataset("iris")
sns.scatterplot(data=df, x="sepal_length", y="sepal_width", hue="species")
plt.show()
```
