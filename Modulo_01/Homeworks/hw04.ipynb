```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("HollywoodMovies.csv")
```

### ¿Que año recaudo más fondos?

#### Entregable

1.- ¿Que muestra su grafica? Muestra las ganancias por año


2.- ¿que patron observan? En todos los años excepto 2009 la evolución es lenta.


3.- ¿Existe algun valor atipico? En 2009 hubo un salto de ganancia que sobresale sobre otros años.


4.- ¿que conclusion pueden obtener? La evolcuión de ganancias que el cine trae va ascendiendo de manera progresiva, sin embargo, 2009 parece haber sido un gran año para el cine logrando reunir casi el doble de presupuesto que sus años anteriores.

```python

plt.bar(df["Year"], df["WorldGross"])
plt.xlabel("Año")
plt.ylabel("Ganancia")
plt.title("Ganancia por año")
plt.show()
```

### Ganancia de cine por año 2007-2013

#### Entregable

1.- ¿Que muestra su grafica? Muestra la evolución de las ganancias por año.


2.- ¿que patron observan? La evolución es ascendente año con año, excepto con una ligera baja en 2010.


3.- ¿Existe algun valor atipico? En 2010 hubo un punto de caída leve pero se recupero.


4.- ¿que conclusion pueden obtener? La evolcuión de ganancias que el cine trae va ascendiendo de manera progresiva, sin embargo, en 2010 hubo una ligera caída que poco después se recupero.

```python
df_lineas = df.groupby("Year")["WorldGross"].sum()

plt.plot(df_lineas.index, df_lineas.values)
plt.xlabel("Año")
plt.ylabel("Ganancia")
plt.title("Ganancia por año")
plt.show()
```

### 2007-2013 Número de Películas por Género

#### Entregable

1.- ¿Que muestra su grafica? Número de películas por género.


2.- ¿que patron observan? Hay muchas más películas de comedia y de acción que de otros géneros.


3.- ¿Existe algun valor atipico? No, comedia y acción siempre han sido los géneros más rentables del cine.


4.- ¿que conclusion pueden obtener? Los géneros de comedia y acción representan el mercado más rentable debido a sus temas fáciles de digerir, otros géneros como musicales y documentales quienes son mayormente consumidos por amantes del género u oyentes casuales, tienen menor número de entregas.

```python
genres = df.groupby("Genre")["Movie"].count()

x = genres.index
y = genres.values
plt.xticks(rotation=45, ha="right")

plt.xlabel("Genre")
plt.ylabel("Movies count")

plt.bar(x, y)
plt.show()
```
