```python
from pandas import DataFrame
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df: DataFrame = pd.read_csv("../irisData.csv")

df.head()
```

## Un histograma de la longitud de un pétalo

```python
sns.histplot(data=df, x="SepalLengthCm", bins=10)
plt.title("Sepal Length Distribution")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

## Un histograma de la longitud de un pétalo por especie

```python
sns.histplot(data=df, x="SepalLengthCm", bins=20, hue="Species", alpha=0.5, palette="viridis")
plt.title("Sepal Length Distribution per Species")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()
```

## Promedio de sépalo por especies

```python
sns.barplot(data=df, x="Species", y="SepalLengthCm", hue="Species", palette="viridis")
plt.title("Sepal Average Distribution per Species")
plt.xlabel("Sepal Length")
plt.ylabel("Specie")
plt.show()
```

## Gráfica de bigotes

```python
sns.boxplot(data=df, x="Species", y="PetalWidthCm", hue="Species", palette="viridis")
plt.title("Mostacho graph :v")
plt.xlabel("Sepal Width ")
plt.ylabel("Specie")
plt.show()
```

# Pairplot

```python
sns.pairplot(data=df, hue="Species", palette="viridis")
plt.show()
```

## Heatmap

```python
plt.figure(figsize=(5, 5))
sns.heatmap(data=df.iloc[:, 0:4].corr(), annot=True, cmap="coolwarm")
plt.tight_layout()
plt.xlabel("Features")
plt.ylabel("Features")
plt.show()
```
